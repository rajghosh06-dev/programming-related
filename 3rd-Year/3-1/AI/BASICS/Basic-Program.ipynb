## Arithmetic Operations

```python
a = 10
b = 20

print('Sum is',a+b)
print('Difference is',a-b)
print('Multiplication is', a*b)
print('Division is',a/b)
```

### Taking Input

```python
a = int(input('A = '))
b = int(input('B = '))

print('Sum is',a+b)
print('Difference is',a-b)
print('Multiplication is', a*b)
print('Division is',a/b)
```

```python
a = float(input('A = '))
b = float(input('B = '))

print('Sum is',a+b)
print('Difference is',a-b)
print('Multiplication is', a*b)
print('Division is',a/b)
```

```python
a = int(input('Enter "a" value: '))
b = int(input('Enter "b" value: '))

print(type(a))
print(type(b))

# Type Casting

a = float(a)
b = b.__float__()

print(type(a))
print(type(b))

print('Sum is',a+b)
print('Difference is',a-b)
print('Multiplication is', a*b)
print('Division is',a/b)
```

## Loop Example - 'while'

```python
i=1
while i<5:
    print('"i" is: ',i)
    i+=1
print('Loop terminated, this is out')
```

## Loop Example - 'for'

```python
for i in range(1,5):
    print('"i" is: ',i)
print('Loop terminated')
```
