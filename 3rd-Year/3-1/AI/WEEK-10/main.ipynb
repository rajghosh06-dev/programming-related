```python
import pandas as pd
# from pgmpy.models import DiscreteBayesianNetwork
# from pgmpy.estimators import MaximumLikelihoodEstimator
# from pgmpy.inference import VariableElimination
from sklearn.naive_bayes import GaussianNB
```

```python
df= pd.read_csv('heart.csv')
print(df)
print('Size of the data-set: ',df.shape)
# x= df.drop(['target'],axis=1)
# print(x)
```

```python
Y=df['target']
X=df.drop(['target'],axis=1)
model=GaussianNB()
model.fit(X,Y)  # inference taking new patient data

age = int(input('Enter age of the patient: '))
sex = int(input('Enter sex of the patient: '))
cp = int(input('Enter cp of the patient: '))

# result=model.predict([50,1,2])
inp = pd.DataFrame([[age, sex, cp]], columns=['age', 'sex', 'cp'])
result=model.predict(inp)

if(result==1):
    print('the patient has heart disease')
else:
    print('the patient has no heart disease')
```
