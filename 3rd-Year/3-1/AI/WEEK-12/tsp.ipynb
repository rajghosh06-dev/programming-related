```python
#TSP:
n = int(input('Enter the number of cities:'))
city = {}
w = []
for _ in range(n):
    cit = int(input(f'Enter City {_+1}:'))
    w = map(int, (input(f'Enter Weight {_+1}:')).split())
    city[cit].update(w)

print(city)

```

```python
## [SIR]
# TSP using DFS and Backtracking

graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(graph)

best_cost = float('inf')
best_path = []

# Visited tracking array
visited = [False] * n
visited[0] = True  # Start at city 0

# DFS Function
def tsp(city, count, cost, path):
    global best_cost, best_path

    # Pruning: stop early if current cost exceeds the best found so far
    if cost >= best_cost:
        return

    # Base case: If all cities are visited
    if count == n:
        total_cost = cost + graph[city][0]  # Return to start city (0)
        if total_cost < best_cost:
            best_cost = total_cost
            best_path = path + [0]
        return

    # Try every unvisited city
    for i in range(n):
        if not visited[i]:
            visited[i] = True  # Choose (Visit City)

            tsp(i, count + 1, cost + graph[city][i], path + [i])  # Explore

            visited[i] = False  # Backtrack (Unvisit City)

# Execute starting from city 0
tsp(0, 1, 0, [0])

print('Best Route:', best_path)
print('Minimum Distance:', best_cost)
```
