# WEEK-2 (A): List Operation Program

```python
# TYPE-1

# 1. Initialization & Nested List
L = [10, 20, 30, [100, 200]]
print("Original List:", L)
print("Nested element (L[3][1]):", L[3][1])

# 2. Length
print("Length of L:", len(L))

# 3. Concatenation
L2 = [40, 50]
print("Concatenation (L + L2):", L + L2)

# 4. Membership
print("Is 20 in L?", 20 in L)
print("Is 99 in L?", 99 in L)

# 5. Iteration
print("Iteration elements: ", end="")
for item in L2:
    print(item, end=" ")
print() # newline

# 6. Indexing and Slicing
print("Indexing (L[1]):", L[1])
print("Slicing (L[0:2]):", L[0:2])

# 7. Add: Append, Insert, and Extend
L.append(60)         # Adds 60 to the end
L.insert(1, 15)      # Inserts 15 at index 1
L.extend([70, 80])   # Adds multiple elements to the end
print("After Append, Insert, Extend:", L)

# 8. Delete: Remove, Pop, and Del
L.remove(30)         # Removes the first occurrence of 30
popped_val = L.pop() # Removes and returns the last element
del L[0]             # Deletes the element at index 0
print("After Deletions:", L)
```

```python
## User Defined Functions
L = []
M = [100,200,300,400,500]


def printList():
    print('LIST "L": ',L)

def appendOpp():
    print('-- APPEND OPERATION --')
    el = int(input('Enter element to add in List:'))
    L.append(el)
    printList()

def insertOpp():
    print('-- APPEND OPERATION --')
    el = int(input('Enter element to add in List:'))
    ind = int(input('Enter index position to insert:'))
    L.insert(ind,el)
    printList()

def accessOpp():
    print('-- ACCESS OPERATION --')
    ind = int(input('Enter position to access from List:'))
    print(f'Value at index {ind} is: ',L[ind])

def updateOpp():
    print('-- UPDATE OPERATION --')
    update_ind = int(input('Enter position to UPDATE from List:'))
    e6 = int(input(f'Enter element to update at position {update_ind} in List:'))
    L[update_ind] = e6
    print(f'List after update at pos {update_ind} is: ',L)

def deleteOpp():
    print('-- DELETE OPERATION --')
    del_ind = int(input('Enter position to DELETE from List:'))
    del L[del_ind]
    print(f'List after delete at pos {del_ind} is: ',L)

def repeatOpp():
    print('-- REPEAT OPERATION --')
    repeat_ind = int(input('Enter how many times to REPEAT List:'))
    print(L*repeat_ind)

def minOpp():
    print('-- MIN OPERATION --')
    print('Minimum value of the List is: ',min(L))

def maxOpp():
    print('-- MAX OPERATION --')
    print('Maximum value of the List is: ',max(L))

def countOpp():
    print('-- COUNT/LENGTH OPERATION --')
    print('Count/Length of the List is: ',len(L))

def mergeOpp():
    print('-- MERGE OPERATION --')
    print('After Merging, the List is: ',L+M)
    print('Original List after Merge Operation is: ',L)

def extendOpp():
    print('-- EXTEND OPERATION --')
    #### Use M as abovee
    L.extend(M)
    print('After Extending, the Original List is: ',L)

def sortAssOpp():
    ### SORT: ASCENDING OPERATION
    print('-- SORT: ASCENDING OPERATION --')
    L.sort()
    print('ASCENDING ORDER of original list: ',L)
def sortDesOpp():
    ### SORT: DESCENDING OPERATION
    print('-- SORT: DESCENDING OPERATION --')
    L.sort(reverse=True)
    print('DESCENDING ORDER of original list: ',L)

def slicingOpp():
    ### SLICING OPERATION
    print('-- SLICING OPERATION --')
    left_ind=int(input('Enter Left Index'))
    right_ind=int(input('Enter Right Index'))
    #### Op1:
    print('SLICING with only Left Index: List',L[left_ind:])
    #### Op2:
    print('SLICING with only Right Index: List',L[right_ind:])
    #### Op3:
    print('SLICING with both Left and Right Index: List',L[left_ind:right_ind])

def membershipOpp():
    ### MEMBERSHIP/SEARCH OPERATION
    print('-- MEMBERSHIP/SEARCH OPERATION --')
    key = int(input('Enter the element you want to find in List: '))
    if key in L:
        print('Element Found!')
    else:
        print('Element Not Found!')

def nestedListOpp():
    ### NESTED-LIST OPERATION
    print('-- NESTED-LIST OPERATION --')
    sub_list = [1000,2000,3000]
    L.append(sub_list)
    print('NESTED-LIST is: ',L)
```

```python
appendOpp()
insertOpp()
accessOpp()
updateOpp()
deleteOpp()
repeatOpp()
minOpp()
maxOpp()
countOpp()
mergeOpp()
extendOpp()
sortAssOpp()
sortDesOpp()
slicingOpp()
membershipOpp()
nestedListOpp()
```
